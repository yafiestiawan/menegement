```python

# ==================================================
#           AUTOMATED TELLER MACHINE (ATM)
#                  ATM SYSTEM V1.0
# ==================================================

from datetime import datetime

# ---------------- ACCOUNT INFORMATION ----------------

account_holder = "John Doe"
account_number = "1234567890"
pin = "1234"
balance = 5_000_000

transaction_history = []
login_attempts = 0
maximum_attempts = 3


# ---------------- WELCOME SCREEN ----------------

def display_welcome():
    print("\n" + "=" * 50)
    print("          WELCOME TO ATM MACHINE")
    print("          SECURE BANKING SYSTEM")
    print("=" * 50)
    print("Please insert your card to continue.")
    print("")


# ---------------- LOGIN SYSTEM ----------------

def login():
    global login_attempts

    while login_attempts < maximum_attempts:
        entered_pin = input("Please enter your PIN: ")

        if entered_pin == pin:
            print("\nLogin successful!")
            print(f"Welcome, {account_holder}!")
            return True

        login_attempts += 1
        remaining_attempts = maximum_attempts - login_attempts

        print("Incorrect PIN.")

        if remaining_attempts > 0:
            print(f"Attempts remaining: {remaining_attempts}")

    print("\nYour account session has been locked.")
    print("Please contact customer service.")
    return False


# ---------------- ACCOUNT INFORMATION ----------------

def display_account_information():
    print("\n========== ACCOUNT INFORMATION ==========")
    print(f"Account Holder : {account_holder}")
    print(f"Account Number : {account_number}")
    print(f"Current Balance: Rp{balance:,}")
    print("=" * 43)


# ---------------- CHECK BALANCE ----------------

def check_balance():
    print("\n========== BALANCE INQUIRY ==========")
    print(f"Account Holder : {account_holder}")
    print(f"Available Balance: Rp{balance:,}")
    print("=====================================")


# ---------------- TRANSACTION LOGGER ----------------

def record_transaction(transaction_type, amount, details=""):
    transaction = {
        "date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "type": transaction_type,
        "amount": amount,
        "details": details,
        "balance": balance
    }

    transaction_history.append(transaction)


# ---------------- DEPOSIT MONEY ----------------

def deposit_money():
    global balance

    print("\n========== CASH DEPOSIT ==========")

    try:
        amount = int(input("Enter deposit amount: Rp"))

        if amount <= 0:
            print("Error: Amount must be greater than zero.")
            return

        balance += amount

        record_transaction("Deposit", amount)

        print("\nDeposit completed successfully!")
        print(f"Deposited Amount: Rp{amount:,}")
        print(f"Updated Balance : Rp{balance:,}")

    except ValueError:
        print("Error: Please enter a valid whole number.")


# ---------------- WITHDRAW MONEY ----------------

def withdraw_money():
    global balance

    print("\n========== CASH WITHDRAWAL ==========")

    try:
        amount = int(input("Enter withdrawal amount: Rp"))

        if amount <= 0:
            print("Error: Amount must be greater than zero.")
            return

        if amount > balance:
            print("Transaction declined: Insufficient balance.")
            return

        balance -= amount

        record_transaction("Withdrawal", amount)

        print("\nPlease collect your cash.")
        print(f"Withdrawal Amount: Rp{amount:,}")
        print(f"Remaining Balance: Rp{balance:,}")

    except ValueError:
        print("Error: Please enter a valid whole number.")


# ---------------- TRANSFER MONEY ----------------

def transfer_money():
    global balance

    print("\n========== MONEY TRANSFER ==========")

    recipient_account = input(
        "Enter recipient account number: "
    ).strip()

    if not recipient_account.isdigit():
        print("Error: Invalid account number.")
        return

    if recipient_account == account_number:
        print("Error: You cannot transfer money to your own account.")
        return

    try:
        amount = int(input("Enter transfer amount: Rp"))

        if amount <= 0:
            print("Error: Amount must be greater than zero.")
            return

        if amount > balance:
            print("Transaction declined: Insufficient balance.")
            return

        confirmation = input(
            f"Transfer Rp{amount:,} to account "
            f"{recipient_account}? (Y/N): "
        ).strip().upper()

        if confirmation != "Y":
            print("Transfer cancelled.")
            return

        balance -= amount

        record_transaction(
            "Transfer",
            amount,
            f"Recipient: {recipient_account}"
        )

        print("\nTransfer completed successfully!")
        print(f"Recipient Account: {recipient_account}")
        print(f"Transfer Amount  : Rp{amount:,}")
        print(f"Remaining Balance: Rp{balance:,}")

    except ValueError:
        print("Error: Please enter a valid whole number.")


# ---------------- TRANSACTION HISTORY ----------------

def display_transaction_history():
    print("\n========== TRANSACTION HISTORY ==========")

    if not transaction_history:
        print("No transactions have been recorded.")
        return

    for index, transaction in enumerate(
        transaction_history, start=1
    ):
        print(f"\nTransaction #{index}")
        print(f"Date        : {transaction['date']}")
        print(f"Type        : {transaction['type']}")
        print(f"Amount      : Rp{transaction['amount']:,}")
        print(f"Details     : {transaction['details']}")
        print(f"Balance After Transaction: "
              f"Rp{transaction['balance']:,}")

    print("\n=========================================")


# ---------------- CHANGE PIN ----------------

def change_pin():
    global pin

    print("\n========== CHANGE PIN ==========")

    current_pin = input("Enter your current PIN: ")

    if current_pin != pin:
        print("Error: Current PIN is incorrect.")
        return

    new_pin = input("Enter your new 4-digit PIN: ")

    if not new_pin.isdigit() or len(new_pin) != 4:
        print("Error: PIN must contain exactly 4 digits.")
        return

    confirm_pin = input("Confirm your new PIN: ")

    if new_pin != confirm_pin:
        print("Error: PIN confirmation does not match.")
        return

    if new_pin == pin:
        print("Your new PIN must be different.")
        return

    pin = new_pin
    print("Your PIN has been changed successfully.")


# ---------------- MAIN MENU ----------------

def display_menu():
    print("\n" + "=" * 43)
    print("                 ATM MENU")
    print("=" * 43)
    print("1.  Check Balance")
    print("2.  Deposit Money")
    print("3.  Withdraw Money")
    print("4.  Transfer Money")
    print("5.  Transaction History")
    print("6.  Account Information")
    print("7.  Change PIN")
    print("8.  Logout")
    print("=" * 43)


# ---------------- MAIN PROGRAM ----------------

def main():
    display_welcome()

    if not login():
        return

    while True:
        display_menu()

        choice = input("Select an option (1-8): ").strip()

        if choice == "1":
            check_balance()

        elif choice == "2":
            deposit_money()

        elif choice == "3":
            withdraw_money()

        elif choice == "4":
            transfer_money()

        elif choice == "5":
            display_transaction_history()

        elif choice == "6":
            display_account_information()

        elif choice == "7":
            change_pin()

        elif choice == "8":
            print("\nYou have successfully logged out.")
            print("Thank you for using our ATM service.")
            print("Please take your card.")
            print("Have a nice day!")
            break

        else:
            print("Invalid option. Please select 1-8.")


# ---------------- RUN APPLICATION ----------------

if __name__ == "__main__":
    main()
```
